## Population transfer function

```python
#################################################################################################################
## Population transfer function: sigmoid ########################################################################
#################################################################################################################

def sigmoid(xi, beta=1.0, xi0=0.0):
    return 1 / (1 + np.exp(-beta * (xi - xi0)))

#---------------------------------------------------------------------------------------------------------------#
#---------------------------------------------------------------------------------------------------------------#
```

## Model equations

```python
#################################################################################################################
## Model equations ##############################################################################################
#################################################################################################################

## Localized Wilson-Cowan equations
def wilson_cowan(u, t, params):
    E, I = u                                                         # Construct the vector for time integration via odeint                                                 
    wee, wei, wie, wii, P, Q, beta_e, beta_i, xi0_e, xi0_i, mu = params        
    
    # Transfer functions
    Se = sigmoid(wee * E - wei * I + P, beta=beta_e, xi0=xi0_e)       # excitatory population transfer function
    Si = sigmoid(wie * E - wii * I + Q, beta=beta_i, xi0=xi0_i)       # inhibitory population transfer function
    
    # Compute the increments of E and I in time  
    dEdt = 1/mu*(-E + (1-E)*Se)                                                  
    dIdt = 1/mu*(-I + (1-I)*Si)

    
    return [dEdt, dIdt]

#---------------------------------------------------------------------------------------------------------------#
#---------------------------------------------------------------------------------------------------------------#
```

## Model parameters and initial conditions
## Step 1
In the parameters, we notice that P and Q are nonzero. What does that mean? Compare their values to the parameter setting of previous damped oscillations.

```python
#################################################################################################################
##  Model parameters and initial conditions: limit cycles  ######################################################
#################################################################################################################

# Model parameters
params = {
    'wee': 16.0,                             # E → E connection weight from excitatory neurons to excitatory neurons
    'wei': 12.0,                             # I → E connection weight from inhibitory neurons to excitatory neurons
    'wie': 15.0,                             # E → I connection weight from excitatory neurons to inhibitory neurons
    'wii': 3.0,                              # I → I connection weight from inhibitory neurons to inhibitory neurons
    'P': 1.25,                               # External stimulus to the excitatory population
    'Q': 0.5,                                # External stimulus to the inihibitory population
    'beta_e': 1.3, 'beta_i': 2.0,            # Nonlinearity sharpness of the transfer functions
    'xi0_e': 3.0, 'xi0_i': 3.5,              # Activity thresholds of the transfer functions
    'mu': 1.0                                # Time constant
}

# Assign the parameter values for the time integration
param_values = list(params.values())

# Initial conditions for limit cycle: E and I at t=0
y0 = [0.1, 0.1]

#---------------------------------------------------------------------------------------------------------------#
#---------------------------------------------------------------------------------------------------------------#
```

## Step 2
Run the simulation. Comment on the results. How would you interpret these results in the biological context? What is the difference between these results and the simulation result of the resting state and of the damped oscillations from the previous sessions?

```python
#################################################################################################################
## Simulation of the model ######################################################################################
#################################################################################################################

# Time discretization
T = 20                              # final time
nOfSamples = 1000                   # number of time samples
t = np.linspace(0, T, nOfSamples)   # time vector

# Integrate the system in time
sol = odeint(wilson_cowan, y0, t, args=(param_values,))
E, I = sol.T

#---------------------------------------------------------------------------------------------------------------#
#---------------------------------------------------------------------------------------------------------------#


#########################################################################################
## Plot the simulation results ##########################################################
#########################################################################################

# Plot results
min_y = 0; max_y = 1  # For plot range
plt.figure(figsize=(10, 5))
plt.plot(t, E, label='Excitatory (E)', color='b')
plt.plot(t, I, label='Inhibitory (I)', color='r')
plt.ylim(min_y, max_y)
plt.xlabel('Time')
plt.ylabel('Activity')
plt.title('Wilson-Cowan Model Dynamics')
plt.legend()
plt.grid(False)
plt.tight_layout()
plt.show()

#---------------------------------------------------------------------------------------#
#---------------------------------------------------------------------------------------#
```
